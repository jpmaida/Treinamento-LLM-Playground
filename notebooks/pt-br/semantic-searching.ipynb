# Buscas semânticas

Para ser honesto com você, eu considero (minha opinião) a busca semântica a parte mais importante no processo de construção de uma solução de GenAI, e acaba que ela não merece o devido mérito. Os modelos (LLMs) são incríveis devido a capacidadee de raciocínio, mas eles não são mágicos. Eles não sabem tudo, e é aí que entra a busca semântica. Ela é a ponte entre o conhecimento que você tem (ou que você pode acessar) e o modelo que você está usando. Estes concentos ficarão mais claros a frente.

Vamos analisar o seguinte o problema ... imagina que queremos encontrar informações relacionadas à seguinte pergunta:

> **"Quem treinou Luke Skywalker?"**

Uma busca tradicional (baseada em palavras-chave) poderia retornar resultados que contivessem as palavras "Luke" e "Skywalker", mas não necessariamente a resposta correta. Por exemplo, poderia retornar artigos sobre Luke Skywalker na batalha de Hoth, mas não mencionar quem o treinou.

Uma busca tradicional procura palavras exatamente iguais às da consulta.

Por exemplo:

- "Luke Skywalker"
- "treinou"

Se um documento disser:

> "Yoda foi o mestre Jedi responsável pelo treinamento de Luke."

A palavra **"treinou"** não aparece no texto.

Apesar de o documento responder perfeitamente à pergunta, uma busca baseada apenas em palavras-chave pode não encontrá-lo.

Esse é um dos principais problemas da busca lexical: ela compara palavras, mas não compreende o significado por trás delas.

**Como resolve este problema ?**

## Busca léxica vs. Busca semântica

A busca lexical é um método tradicional de recuperação de informações que localiza documentos ou dados comparando palavras, frases ou tokens exatos de uma consulta do usuário com um índice. Em vez de interpretar o significado conceitual ou o contexto de uma frase, ela se baseia na comparação literal de texto e na pontuação com base na frequência dos termos.

Busca semântica (*Semantic Search*) é uma técnica de recuperação de informação que procura documentos com base no **significado** da consulta, e não apenas nas palavras utilizadas.

Ao invés de perguntar:

> "Esse documento contém exatamente esta palavra?"

A busca semântica pergunta:

> "Esse documento possui um significado semelhante ao da consulta?"

Isso permite encontrar documentos relevantes mesmo quando utilizam palavras diferentes, sinônimos ou formas distintas de expressar uma mesma ideia.
